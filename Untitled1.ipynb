```python
# ╔══════════════════════════════════════════════════════════════════╗
# ║   BUĞDAY HASTALIĞI TESPİTİ — TAM PROJE NOTEBOOK                  ║
# ║   SUBU EEF | Yapay Zekaya Giriş | 2024-2025 Bahar                ║
# ╚══════════════════════════════════════════════════════════════════╝
#
# KULLANIM:
#   1. Google Colab'da "Runtime > Change runtime type > T4 GPU" seç
#   2. Her hücreyi sırayla çalıştır (Shift+Enter)
#   3. Kaggle API anahtarı için: kaggle.com/settings > API > Create New Token


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 1 — GPU Kontrolü
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
import tensorflow as tf

gpu = tf.config.list_physical_devices('GPU')
if gpu:
    print("✅ GPU aktif:", gpu[0].name)
else:
    print("⚠️  GPU YOK! Runtime > Change runtime type > T4 GPU seç!")


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 2 — Kütüphaneleri Kur
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
!pip install -q kaggle streamlit pyngrok scikit-learn matplotlib seaborn shap

print("✅ Kütüphaneler kuruldu")


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 3 — Kaggle API Anahtarını Yükle
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# kaggle.com/settings > API > "Create New Token" butonuna bas
# İndirilen kaggle.json dosyasını yükle

from google.colab import files
print("kaggle.json dosyasını seç ve yükle:")
files.upload()

import os
os.makedirs(os.path.expanduser("~/.kaggle"), exist_ok=True)
!cp kaggle.json ~/.kaggle/
!chmod 600 ~/.kaggle/kaggle.json
print("✅ Kaggle API hazır")


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 4 — Veriyi İndir ve Hazırla
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
import os, shutil
import numpy as np

# Veriyi indir (Plant Disease dataset — buğday sınıflarını filtrele)
!kaggle datasets download -d vipoooool/new-plant-diseases-dataset -p /content/raw --unzip -q
print("✅ Veri indirildi")

# Buğday sınıflarını filtrele ve yeni klasör yapısı oluştur
RAW_TRAIN = "/content/raw/New Plant Diseases Dataset(Augmented)/New Plant Diseases Dataset(Augmented)/train"
RAW_VAL   = "/content/raw/New Plant Diseases Dataset(Augmented)/New Plant Diseases Dataset(Augmented)/valid"
DATA_DIR  = "/content/data"

SINIF_MAP = {
    "Wheat___Brown_Rust":        "Kahverengi_Pas",
    "Wheat___Yellow_Rust":       "Sari_Pas",
    "Wheat___Healthy":           "Saglikli",
    "Wheat___Loose_Smut":        "Rastik",
    "Wheat___Septoria":          "Septorya",
    "Wheat___Powdery_Mildew":    "Kulleme",
}

for split, raw_path in [("train", RAW_TRAIN), ("test", RAW_VAL)]:
    for kaynak_ad, hedef_ad in SINIF_MAP.items():
        kaynak = os.path.join(raw_path, kaynak_ad)
        hedef  = os.path.join(DATA_DIR, split, hedef_ad)
        if os.path.exists(kaynak):
            shutil.copytree(kaynak, hedef, dirs_exist_ok=True)
            n = len(os.listdir(hedef))
            print(f"  ✅ {split}/{hedef_ad}: {n} görüntü")

print("\n✅ Veri hazırlandı →", DATA_DIR)


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 5 — Keşifsel Veri Analizi (EDA)
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Sınıf dağılımı
siniflar = os.listdir(os.path.join(DATA_DIR, "train"))
sayilar  = [len(os.listdir(os.path.join(DATA_DIR, "train", s))) for s in siniflar]

plt.figure(figsize=(10, 4))
bars = plt.bar(siniflar, sayilar, color=["#16a34a","#ca8a04","#eab308","#374151","#3b82f6","#f97316"])
plt.title("Sınıf Dağılımı — Eğitim Verisi", fontsize=14)
plt.xticks(rotation=20, ha="right")
plt.ylabel("Görüntü Sayısı")
for bar, n in zip(bars, sayilar):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5, str(n), ha="center", fontsize=10)
plt.tight_layout()
plt.savefig("/content/eda_dagilim.png", dpi=150)
plt.show()

# Örnek görüntüler
fig, axes = plt.subplots(2, 3, figsize=(14, 8))
for ax, sinif in zip(axes.flat, siniflar):
    klasor = os.path.join(DATA_DIR, "train", sinif)
    dosya  = os.path.join(klasor, os.listdir(klasor)[0])
    img    = mpimg.imread(dosya)
    ax.imshow(img)
    ax.set_title(sinif.replace("_", " "), fontsize=11)
    ax.axis("off")
plt.suptitle("Her Sınıftan Örnek Görüntüler", fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig("/content/eda_ornekler.png", dpi=150)
plt.show()

print("✅ EDA tamamlandı")


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 6 — Veri Generatörlerini Hazırla
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
from tensorflow.keras.preprocessing.image import ImageDataGenerator

IMG_SIZE   = 224
BATCH_SIZE = 32
NUM_CLASSES = len(siniflar)

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    horizontal_flip=True,
    zoom_range=0.1,
    brightness_range=[0.8, 1.2],
    validation_split=0.2
)
test_datagen = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    os.path.join(DATA_DIR, "train"),
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="training"
)
val_gen = train_datagen.flow_from_directory(
    os.path.join(DATA_DIR, "train"),
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="validation"
)
test_gen = test_datagen.flow_from_directory(
    os.path.join(DATA_DIR, "test"),
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

print(f"✅ Eğitim: {train_gen.samples} | Doğrulama: {val_gen.samples} | Test: {test_gen.samples}")
print("Sınıflar:", train_gen.class_indices)


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 7 — Baseline Model (sklearn — Random Forest)
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report
from PIL import Image

def goruntu_yukle(klasor, max_per_class=100, size=64):
    X, y = [], []
    for label, sinif in enumerate(sorted(os.listdir(klasor))):
        sinif_yolu = os.path.join(klasor, sinif)
        dosyalar   = os.listdir(sinif_yolu)[:max_per_class]
        for dosya in dosyalar:
            try:
                img = Image.open(os.path.join(sinif_yolu, dosya)).convert("RGB").resize((size, size))
                X.append(np.array(img).flatten() / 255.0)
                y.append(label)
            except:
                pass
    return np.array(X), np.array(y)

print("Baseline için görüntüler yükleniyor (64x64, hızlı)...")
X_train, y_train = goruntu_yukle(os.path.join(DATA_DIR, "train"), max_per_class=80)
X_test,  y_test  = goruntu_yukle(os.path.join(DATA_DIR, "test"),  max_per_class=20)

print(f"Eğitim: {X_train.shape}, Test: {X_test.shape}")

# Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
rf_acc = rf.score(X_test, y_test)
print(f"\n🌲 Random Forest Doğruluğu: {rf_acc*100:.1f}%")

# SVM
svm = SVC(kernel="rbf", C=1.0, random_state=42)
svm.fit(X_train, y_train)
svm_acc = svm.score(X_test, y_test)
print(f"🔷 SVM Doğruluğu: {svm_acc*100:.1f}%")

print("\n(CNN ile bunları çok geçeceğiz!)")


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 8 — CNN Model Mimarisi (EfficientNetB3 Transfer Learning)
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import EfficientNetB3

base_model = EfficientNetB3(weights="imagenet", include_top=False, input_shape=(IMG_SIZE, IMG_SIZE, 3))
base_model.trainable = False  # Faz 1: dondurulmuş

inputs  = keras.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
x       = base_model(inputs, training=False)
x       = layers.GlobalAveragePooling2D()(x)
x       = layers.Dropout(0.3)(x)
x       = layers.Dense(128, activation="relu")(x)
x       = layers.Dropout(0.2)(x)
outputs = layers.Dense(NUM_CLASSES, activation="softmax")(x)

model = keras.Model(inputs, outputs)
model.compile(
    optimizer=keras.optimizers.Adam(1e-4),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
model.summary()
print("✅ Model oluşturuldu")


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 9 — Faz 1: Feature Extraction Eğitimi
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
os.makedirs("/content/models", exist_ok=True)
os.makedirs("/content/docs", exist_ok=True)

callbacks = [
    keras.callbacks.EarlyStopping(patience=5, restore_best_weights=True, verbose=1),
    keras.callbacks.ReduceLROnPlateau(patience=3, factor=0.5, verbose=1),
    keras.callbacks.ModelCheckpoint("/content/models/bugday_model.h5", save_best_only=True, verbose=1),
]

print("─── Faz 1: Feature Extraction (20 epoch max) ───")
history1 = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=20,
    callbacks=callbacks,
    verbose=1
)
print(f"✅ Faz 1 tamamlandı. En iyi val_accuracy: {max(history1.history['val_accuracy'])*100:.1f}%")


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 10 — Faz 2: Fine-Tuning
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
base_model.trainable = True
for layer in base_model.layers[:-30]:
    layer.trainable = False

model.compile(
    optimizer=keras.optimizers.Adam(1e-5),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("─── Faz 2: Fine-Tuning (son 30 katman, 30 epoch max) ───")
history2 = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=30,
    callbacks=callbacks,
    verbose=1
)
print(f"✅ Faz 2 tamamlandı. En iyi val_accuracy: {max(history2.history['val_accuracy'])*100:.1f}%")


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 11 — Model Değerlendirme & Metrikler
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix

best_model = keras.models.load_model("/content/models/bugday_model.h5")

y_pred_prob = best_model.predict(test_gen)
y_pred      = np.argmax(y_pred_prob, axis=1)
y_true      = test_gen.classes
sinif_adlari = list(test_gen.class_indices.keys())

print("─── Classification Report ───")
print(classification_report(y_true, y_pred, target_names=sinif_adlari))

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt="d", cmap="Greens",
            xticklabels=sinif_adlari, yticklabels=sinif_adlari)
plt.title("Confusion Matrix — Buğday Hastalığı Tespiti")
plt.ylabel("Gerçek")
plt.xlabel("Tahmin")
plt.tight_layout()
plt.savefig("/content/docs/confusion_matrix.png", dpi=150)
plt.show()
print("✅ Confusion matrix kaydedildi")


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 12 — Eğitim Grafiği
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

e1 = range(1, len(history1.history["accuracy"]) + 1)
e2 = range(len(e1) + 1, len(e1) + len(history2.history["accuracy"]) + 1)

axes[0].plot(e1, history1.history["accuracy"],     "b-o", ms=4, label="Eğitim Faz1")
axes[0].plot(e1, history1.history["val_accuracy"], "b--", ms=4, label="Doğrulama Faz1")
axes[0].plot(e2, history2.history["accuracy"],     "g-o", ms=4, label="Eğitim Faz2")
axes[0].plot(e2, history2.history["val_accuracy"], "g--", ms=4, label="Doğrulama Faz2")
axes[0].set_title("Doğruluk"); axes[0].legend(); axes[0].grid(alpha=0.3)

axes[1].plot(e1, history1.history["loss"],     "b-o", ms=4, label="Eğitim Faz1")
axes[1].plot(e1, history1.history["val_loss"], "b--", ms=4, label="Doğrulama Faz1")
axes[1].plot(e2, history2.history["loss"],     "g-o", ms=4, label="Eğitim Faz2")
axes[1].plot(e2, history2.history["val_loss"], "g--", ms=4, label="Doğrulama Faz2")
axes[1].set_title("Kayıp"); axes[1].legend(); axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig("/content/docs/egitim_grafigi.png", dpi=150)
plt.show()
print("✅ Eğitim grafiği kaydedildi")


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 13 — Modeli Google Drive'a Kaydet (kaybolmasın!)
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
from google.colab import drive
drive.mount("/content/drive")

import shutil
os.makedirs("/content/drive/MyDrive/bugday_projesi", exist_ok=True)

shutil.copy("/content/models/bugday_model.h5",        "/content/drive/MyDrive/bugday_projesi/bugday_model.h5")
shutil.copy("/content/docs/confusion_matrix.png",     "/content/drive/MyDrive/bugday_projesi/confusion_matrix.png")
shutil.copy("/content/docs/egitim_grafigi.png",       "/content/drive/MyDrive/bugday_projesi/egitim_grafigi.png")
shutil.copy("/content/eda_dagilim.png",               "/content/drive/MyDrive/bugday_projesi/eda_dagilim.png")
shutil.copy("/content/eda_ornekler.png",              "/content/drive/MyDrive/bugday_projesi/eda_ornekler.png")

print("✅ Model ve grafikler Google Drive'a kaydedildi!")
print("   📁 Google Drive > bugday_projesi klasörünü kontrol et")


# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# HÜCRE 14 — Streamlit Demo'yu Çalıştır
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# ÖNCE: ngrok.com adresine gidip ücretsiz kayıt ol
# Sonra: dashboard.ngrok.com/tunnels/authtokens adresinden token kopyala

NGROK_TOKEN = "BURAYA_NGROK_TOKEN_YAZ"  # ← bunu değiştir

!pip install pyngrok -q

# app.py dosyasını GitHub'dan çek
!wget -q https://raw.githubusercontent.com/Hilmisngl/bugday-hastaligi-tespiti/main/app.py -O /content/app.py
print("✅ app.py indirildi")

# Modeli app.py'nin bulabileceği yere kopyala
os.makedirs("/content/models", exist_ok=True)

import subprocess, threading, time
from pyngrok import ngrok

ngrok.set_auth_token(NGROK_TOKEN)

def run():
    subprocess.run(["streamlit", "run", "/content/app.py",
                    "--server.port=8501", "--server.headless=true"])

threading.Thread(target=run, daemon=True).start()
time.sleep(4)

url = ngrok.connect(8501)
print("\n" + "="*50)
print("🚀 DEMO CANLI!")
print("🔗 Linkin:", url)
print("="*50)
print("Bu linki tarayıcında aç veya hocana gönder!")
```
