```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# İndirilen ana dizinin içindeki asıl veri klasörüne giriyoruz
asil_veri_yolu = os.path.join(path, "DataSet_New200")

siniflar = []
sayilar = []
toplam_goruntu = 0

print("Veri setindeki sınıflar ve görüntü sayıları:\n")
print("-" * 40)

# Sınıfları ve fotoğraf sayılarını tespit etme
for sinif_adi in sorted(os.listdir(asil_veri_yolu)):
    sinif_yolu = os.path.join(asil_veri_yolu, sinif_adi)

    if os.path.isdir(sinif_yolu):
        goruntu_sayisi = len(os.listdir(sinif_yolu))
        siniflar.append(sinif_adi)
        sayilar.append(goruntu_sayisi)
        toplam_goruntu += goruntu_sayisi
        print(f"{sinif_adi}: {goruntu_sayisi} fotoğraf")

print("-" * 40)
print(f"Toplam fotoğraf sayısı: {toplam_goruntu}\n")

# Veri dağılımını görselleştirme (Sütun Grafiği)
plt.figure(figsize=(12, 6))
sns.barplot(x=sayilar, y=siniflar, hue=siniflar, palette="viridis", legend=False)
plt.title("Buğday Veri Seti - Sınıf Dağılımı")
plt.xlabel("Fotoğraf Sayısı")
plt.ylabel("Buğday Türleri/Sınıfları")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.image as mpimg

# Her sınıftan 1 örnek fotoğraf çizdirelim
plt.figure(figsize=(15, 10))

for i, sinif in enumerate(siniflar):
    sinif_yolu = os.path.join(asil_veri_yolu, sinif)
    # Sınıfın içindeki ilk fotoğrafın adını alıyoruz
    ornek_resim_adi = os.listdir(sinif_yolu)[0]
    resim_yolu = os.path.join(sinif_yolu, ornek_resim_adi)

    # Fotoğrafı okuyup ekrana yansıtıyoruz
    img = mpimg.imread(resim_yolu)
    plt.subplot(3, 4, i + 1) # 3 satır, 4 sütunluk bir ızgara
    plt.imshow(img)
    plt.title(sinif, fontsize=10)
    plt.axis("off") # Eksen çizgilerini kapat

plt.suptitle("Buğday Hastalıkları / Türleri Örnek Fotoğraflar", fontsize=16)
plt.tight_layout()
plt.show()
```
